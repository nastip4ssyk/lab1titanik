```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv("titanic.csv")
print("Размер датасета:", df.shape)
df.head()
```

```python
df.info()
```

```python
print("Пропущенные значения по столбцам:")
print(df.isnull().sum())
```

```python
# Embarked — категориальный, заполняем модой
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Age — числовой, заполняем медианой
df['Age'] = df['Age'].fillna(df['Age'].median())

# Cabin — слишком много пропусков (687 из 891), удаляем столбец
df = df.drop('Cabin', axis='columns')

print("После заполнения:")
print(df.isnull().sum())
```

```python
scaler = MinMaxScaler()

df['Age'] = scaler.fit_transform(df[['Age']])
df['Fare'] = scaler.fit_transform(df[['Fare']])

print("После нормализации:")
df[['Age', 'Fare']].head(10)
```

```python
df = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)
df = df.drop(['Name', 'Ticket', 'PassengerId'], axis='columns')

print("Финальный датасет:", df.shape)
df.head()
```

```python
df.to_csv("processed_titanic.csv", index=False)
print("Сохранено: processed_titanic.csv")
```
